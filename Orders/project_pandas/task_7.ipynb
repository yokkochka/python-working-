```python
import pandas as pd
```

```python
df = pd.read_csv("./csv/clear_data.csv")
df
```

```python
df = df.sort_values(["Country", "Year"])
df
```

```python
df_future = df[["Country", "Year", "Population"]].copy()
df_future
```

```python
df_future["Year"] = df_future["Year"] - 3
df_future
```

```python
df_future = df_future.rename(columns={"Population":"Population_3y"})
df_future
```

```python
df_merged = df.merge(df_future, on = ["Country", "Year"], how = "left")
# df_merged.to_csv("./csv/temp.csv")
df_merged
```

```python
df
```

```python
df_merged= df_merged.dropna(subset=["Population_3y"])
df_merged
```

```python
df_merged["Growth_3y"] = df_merged["Population_3y"] - df_merged["Population"]
df_merged
```

```python
idx_max = df_merged["Growth_3y"].idxmax()
idx_max
```

```python
best_row = df_merged.loc[idx_max]
best_row
```

```python
best_country = best_row["Country"]
best_start_year = int(best_row["Year"])
best_end_year = best_start_year + 3
best_growth = best_row["Growth_3y"]

print(f"Страна: {best_country} ({best_start_year} - {best_end_year}) c приростом населения в {best_growth}")
```
