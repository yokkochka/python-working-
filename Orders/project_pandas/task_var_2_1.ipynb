```python
import pandas as pd


df = pd.read_csv('./csv/clear_data.csv')
df
```

```python
china_india = df[df['Country'].isin(['China', 'India'])]
china_india
```

```python
pivot = china_india.pivot(index='Year', columns='Country', values='Growth Rate (%)')
pivot
```

```python
pivot['diff'] = (pivot['China'] - pivot['India']).abs()
pivot
```

```python
max_diff_year = pivot[pivot['diff'] == pivot['diff'].max()]

print(max_diff_year)
```
