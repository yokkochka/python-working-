```python
import pandas as pd
```

```python
df = pd.read_csv("./csv/clear_data.csv")
df
```

```python
max_year = df['Year'].idxmax()
max_year = df.loc[max_year, "Year"]
print(max_year)
# print(max_country)

min_year = max_year - 10


print(min_year) 
```

```python
last_10_year = df[df["Year"].between(min_year, max_year)]
last_10_year

last_10_year = last_10_year.sort_values("Year")
last_10_year

```

```python
change = (last_10_year.groupby("Country").agg(start = ("Population", "first"), end = ("Population", "last")))
change
```

```python
change['Growth %'] = (change['end'] - change['start']) / change['start'] * 100
change
```

```python
top3 = change.sort_values("Growth %", ascending=False).head(3)

top3
```

```python
df = pd.read_csv('./csv/clear_data.csv')
df
```

```python
max_year = df["Year"].idxmax()
max_year= df.loc[max_year, "Year"]

min_year = max_year - 10

print(max_year)
print(min_year)
```

```python
last_10_year = df[df['Year'].between(min_year, max_year)]
last_10_year
```

```python
change = last_10_year.groupby('Country').agg(start = ("Population", "first"), end = ("Population", "last"))
change
```

```python
change['diff'] = change['end'] - change['start']
change
```

```python
top_3 = change.sort_values("diff").head(3)

top_3
```
