```python
import pandas as pd
```

```python
df = pd.read_csv("./csv/clear_data.csv")
```

```python
df_2007 = df[df["Year"] == 2007]
df_2007


df_2007 = df_2007[~df_2007["Country"].isin(["World", "Middle income", "IBRD only", "Upper middle income", "lower middle income", "Low & middle income", "IDA & IBRD total"])]
```

```python
top5 = df_2007.sort_values("Population", ascending=False).head(5)
top5
```

```python
world_population = df_2007["Population"].sum()
world_population
```

```python
top5["Share (%)"] = (top5["Population"] / world_population) * 100
top5
```

```python
d =  100 - top5["Share (%)"].sum()
d
```

```python
world_population
```

```python
top5.loc[5] = ["OTHER", ' ', 2007, world_population, 0,  0,  '2000s', d]
top5

```

```python
top5.plot(
    kind = "pie",
    y = "Share (%)",
    labels = top5["Country"],
    figsize = (8, 8),
    autopct = "%1.1f%%"



)
```
