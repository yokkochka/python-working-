```python

import pandas as pd

df = pd.read_csv("./csv/clear_data.csv")

df
```

```python
target_col = "Growth Rate (%)"

df["growth Rate (%) abs"] = df[target_col].abs()

df

```

```python
max_index = df['growth Rate (%) abs'].idxmax()

print(max_index)
```

```python
max_coutry = df.loc[max_index, "Country"]

print(max_coutry)
```

```python


# df

target_data= df[df['Country'] == max_coutry]

target_data.head()

```

```python
target_data.plot(
    x = 'Year',
    y = target_col,
    linestyle = "--",
    color = "green",
    label = max_coutry,
    ylabel = "Growth Rate (%)",
    xlabel= 'Year',


    title = f"Прирост населения страны {max_coutry} (%)",

)
```
