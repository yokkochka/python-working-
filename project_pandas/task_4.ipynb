```python
import pandas as pd

df = pd.read_csv("./csv/clear_data.csv")
df
```

```python
mask_a = df["Country"].str.startswith("A") 
mask_a


```

```python

country_with_a = df[mask_a]
country_with_a
```

```python
country_with_a_and_2000_year = country_with_a[country_with_a["Year"] == 2000]
country_with_a_and_2000_year
```

```python
max_index = country_with_a_and_2000_year["Population"].idxmax()
country_max = country_with_a_and_2000_year.loc[max_index, "Country"]

print(max_index)
print(country_max)
```

```python
min_index = country_with_a_and_2000_year["Population"].idxmin()
country_min = country_with_a_and_2000_year.loc[min_index, "Country"]

print(min_index)
print(country_min)

```

```python
year_mask = (df["Year"] >= 1995) & (df["Year"] <= 2005)
year_mask
```

```python
max_pop_data = df[(df["Country"] == country_max) & (year_mask)]
max_pop_data
```

```python

min_pop_data = df[(df["Country"] == country_min) & (year_mask)]
min_pop_data
```

```python
bx = max_pop_data.plot(
    x = "Year",
    y = "Population", 
    kind = "line",
    color = "blue",
    linestyle = "-",
    marker = "o",
    label = country_max,
    title = "Динамика начеления 1995-2005"
)

min_pop_data.plot(
    x = "Year",
    y = "Population", 
    kind = "line",
    color = "green",
    linestyle = "-",
    marker = "o",
    label = country_min,
    ax=bx   
)

bx.legend()
bx.set_ylabel("Population")
bx.grid(True, linestyle = ":", alpha = 0.5)

```
