```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data.csv")

df_numeric = df.drop(columns=["№", "ФИ"])
# df_numeric = df
df_numeric 
```

```python
corr_matrix = df_numeric.corr(method="spearman")
print(corr_matrix)
```

```python
# import numpy as np
# import matplotlib.pyplot as plt
# import seaborn as sns

# # Форматируем значения
# labels = np.array([
#     [f"{x:.2f}".replace('.', ',') for x in row]
#     for row in corr_matrix.values
# ])

# plt.figure(figsize=(12, 10))

# sns.heatmap(
#     corr_matrix,
#     annot=labels,   # свои подписи
#     fmt="",         # формат не нужен
#     cmap="coolwarm",
#     square=True
# )

# plt.title("Spearman Correlation Matrix")
# plt.show()


# import numpy as np
# import matplotlib.pyplot as plt
# import seaborn as sns
# from matplotlib.ticker import FuncFormatter

# # Форматируем значения внутри heatmap
# labels = np.array([
#     [f"{x:.2f}".replace('.', ',') for x in row]
#     for row in corr_matrix.values
# ])

# plt.figure(figsize=(12, 10))

# ax = sns.heatmap(
#     corr_matrix,
#     annot=labels,
#     fmt="",
#     cmap="coolwarm",
#     square=True
# )

# # Замена точек на запятые на цветовой шкале
# cbar = ax.collections[0].colorbar
# cbar.ax.yaxis.set_major_formatter(
#     FuncFormatter(lambda x, pos: f"{x:.1f}".replace('.', ','))
# )

# # plt.title("Spearman Correlation Matrix")
# plt.show()



import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# Шрифт Times New Roman для всего графика
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 12

# Форматируем значения внутри heatmap
labels = np.array([
    [f"{x:.2f}".replace('.', ',') for x in row]
    for row in corr_matrix.values
])

plt.figure(figsize=(12, 10))

ax = sns.heatmap(
    corr_matrix,
    annot=labels,
    fmt="",
    cmap="coolwarm",
    square=True,
    annot_kws={
        "fontsize": 10,
        "fontfamily": "Times New Roman"
    }
)

# Заголовок
plt.title("Корреляционная матрица по Спирмену", fontsize=16)

# Подписи осей
ax.set_xticklabels(
    ax.get_xticklabels(),
    rotation=45,
    ha='right',
    fontname='Times New Roman',
    fontsize=12
)

ax.set_yticklabels(
    ax.get_yticklabels(),
    rotation=0,
    fontname='Times New Roman',
    fontsize=12
)

# Формат цветовой шкалы
cbar = ax.collections[0].colorbar
cbar.ax.yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x:.1f}".replace('.', ','))
)

for label in cbar.ax.get_yticklabels():
    label.set_fontname('Times New Roman')
    label.set_fontsize(12)

plt.tight_layout()
plt.show()
```
